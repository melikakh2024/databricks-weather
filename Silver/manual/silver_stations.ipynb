```python
from pyspark.sql.functions import col
df=spark.table("workspace.bronzeweather.bronze_stations")
df=df.select(    
    "filepath",
    "modificationtime",
    "ingesttime",
    "ID",
    "LATITUDE",
    "LONGITUDE",
    "ELEVATION")
df=df.withColumn("LATITUDE",col("LATITUDE").cast("double")).withColumn("LONGITUDE",col("LONGITUDE").cast("double"))\
.withColumn("ELEVATION",col("ELEVATION").cast("double"))
df.write.option("overwriteschema","true").mode('overwrite').saveAsTable('workspace.silverweather.states_clean')
```
