```python
df=spark.readStream.table("workspace.bronzeweather.years_bronzestream")
from pyspark.sql.functions import col,to_date
df=df.filter(col('Q_FLAG').isNull())
df=df.select("filepath","modificationtime","ingesttime","ID","DATE","ELEMENT","DATA_VALUE")
df=df.withColumn("DATE",to_date("DATE","yyyyMMdd")).withColumn("DATA_VALUE",col("DATA_VALUE").cast("int"))
df.writeStream.outputMode("append").trigger(availableNow=True).option("checkpointLocation","/Volumes/workspace/default/weatherproject/checkpointsilver").toTable("workspace.silverweather.years_clean")
```
