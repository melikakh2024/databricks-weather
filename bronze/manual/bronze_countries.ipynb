```python

from pyspark.sql import functions as sf

df_bronze = (
    spark.read
    .format("text")
    .load("/Volumes/workspace/default/weatherproject/countries.txt")
)


df_bronze = df_bronze.withColumn("CODE", sf.trim(sf.substring("value", 1, 2)))
df_bronze = df_bronze.withColumn("NAME", sf.trim(sf.substring("value", 4, 100)))


df_bronze = df_bronze.select(
    sf.col("_metadata.file_path").alias("filepath"),
    sf.col("_metadata.file_modification_time").alias("modificationtime"),
    sf.current_timestamp().alias("ingesttime"),
    "CODE",
    "NAME"
)


df_bronze.write.format("delta").mode("overwrite").option("overwriteSchema", "true") \
    .saveAsTable("workspace.bronzeweather.bronze_countries")
```
