```python
Python
import requests
from pathlib import Path


path = Path(".")

url_countries = "https://www.ncei.noaa.gov/pub/data/ghcn/daily/ghcnd-countries.txt"
file_countries = path / "countries.txt"
meta_countries = path / "countries_last_modified.txt"

# Get metadata header from NOAA server
head_resp = requests.head(url_countries)
head_resp.raise_for_status()
server_last_modified = head_resp.headers.get("Last-Modified")

# Read previous modification timestamp if local file exists
local_last_modified = ""
if meta_countries.is_file() and meta_countries.stat().st_size > 0:
    with open(meta_countries, "r") as f:
        local_last_modified = f.read().strip()

# Conditional download check
if (not file_countries.is_file()) or (server_last_modified != local_last_modified):
    print("Downloading updated countries file...")
    resp = requests.get(url_countries)
    resp.raise_for_status()
    # Save primary dataset
    with open(file_countries, "wb") as f:
        f.write(resp.content)
    # Save server timestamp to metadata file
    if server_last_modified:
        with open(meta_countries, "w") as f:
            f.write(server_last_modified)
    print("Countries file downloaded successfully.")
else:
    print("Countries file is up-to-date. Download skipped.")
```
